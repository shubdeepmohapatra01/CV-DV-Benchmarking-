```python
%reload_ext autoreload
%autoreload 2

from functools import partial
from custom_gates import bosonic_qaoa
from scipy.optimize import minimize
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define fixed parameters
cutoff = 32
a = 3.5
p = 10
s = 1.0
n = 2   

costval = []
estval = []

def cost_function(params):
    return bosonic_qaoa.cvQAOA(params, cutoff, p, s, n, a, costval, estval)

result = minimize(
    cost_function,
    [6.25]*20,   
    method="SLSQP",
    tol=1e-10,
    options={'maxiter': 5000}
)
```

```python
final_params = result.x

minima,_,_ = bosonic_qaoa.results_final(final_params, cutoff, p, s, n, a)

print(f'Minima found at {minima}')
```
